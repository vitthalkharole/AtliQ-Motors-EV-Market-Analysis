```python
#Basic EDA
total_ev_sales=maker_df["electric_vehicles_sold"].sum()
print("total_ev_sales:",total_ev_sales)
```

```python
#sales by vehicle category
sales_by_category = maker_df.groupby("vehicle_category")["electric_vehicles_sold"].sum()
print("sales by category:",sales_by_category)
```

```python
#Ev sales by fiscal year
maker_df = maker_df.merge(
    date_df,
    on="date",
    how="left"
)
```

```python
yearly_ev_sales = maker_df.groupby(["fiscal_year","vehicle_category"])["electric_vehicles_sold"].sum().reset_index()
yearly_ev_sales
```

```python
yearly_ev_sales["growth_%"] = yearly_ev_sales.groupby("vehicle_category")["electric_vehicles_sold"].pct_change()*100
yearly_ev_sales
```
